#### Necesito que desarrolles un programa en Python usando Programación Orientada a Objetos (POO). Es importante que no utilices manejo de errores (try/except) ni validación de datos.

El programa debe basarse en una clase llamada "Comision_Bancaria".

Requisitos de la clase:
Debe tener exactamente 5 atributos relacionados con una comisión bancaria. Puedes usar o mejorar estos ejemplos:
numeroCuenta
saldo
tipoComision
porcentajeComision
historialTransacciones
También debe incluir exactamente 5 funciones:
agregar_saldo()
retirar_dinero()
calcular_comision()
mostrar_datos()
aplicar_comision()
Requisitos del programa:
Debe ser interactivo, usando un menú en consola con while, if, elif y else.
El usuario debe poder:
Agregar saldo
Retirar dinero
Consultar información
Aplicar comisión
Salir del sistema
Condiciones del código:
Debe estar bien organizado
Incluir comentarios explicativos
Ser fácil de entender para estudiantes
Usar nombres claros en español
No usar librerías externas
No usar validación de datos ni manejo de errores

```python
class Comision_Bancaria:

    # Ahora tiene 6 atributos
    def __init__(self, nombreTitular, numeroCuenta, saldo, tipoComision, porcentajeComision):
        self.nombreTitular = nombreTitular  # NUEVO atributo
        self.numeroCuenta = numeroCuenta
        self.saldo = saldo
        self.tipoComision = tipoComision
        self.porcentajeComision = porcentajeComision
        self.historialTransacciones = []

    def agregar_saldo(self, cantidad):
        self.saldo += cantidad
        self.historialTransacciones.append(f"Ingreso: +${cantidad}")

    def retirar_dinero(self, cantidad):
        self.saldo -= cantidad
        self.historialTransacciones.append(f"Retiro: -${cantidad}")

    def calcular_comision(self):
        comision = self.saldo * (self.porcentajeComision / 100)
        return comision

    def aplicar_comision(self):
        comision = self.calcular_comision()
        self.saldo -= comision
        self.historialTransacciones.append(f"Comisión aplicada: -${comision}")

    def mostrar_datos(self):
        print("\n--- DATOS DE LA CUENTA ---")
        print("Titular:", self.nombreTitular)  # Mostrar nuevo atributo
        print("Número de cuenta:", self.numeroCuenta)
        print("Saldo actual:", self.saldo)
        print("Tipo de comisión:", self.tipoComision)
        print("Porcentaje de comisión:", self.porcentajeComision, "%")
        print("\nHistorial de transacciones:")
        for movimiento in self.historialTransacciones:
            print("-", movimiento)


# ----------- INGRESO DE DATOS -----------

print("=== REGISTRO DE CUENTA BANCARIA ===")

nombre = input("Nombre del titular: ")  # NUEVO
numero = input("Número de cuenta: ")
saldo_inicial = float(input("Saldo inicial: "))
tipo = input("Tipo de comisión (Mensual/Anual): ")
porcentaje = float(input("Porcentaje de comisión (%): "))

# Crear objeto
cuenta = Comision_Bancaria(nombre, numero, saldo_inicial, tipo, porcentaje)


# ----------- MENÚ -----------

while True:
    print("\n===== MENÚ =====")
    print("1. Agregar saldo")
    print("2. Retirar dinero")
    print("3. Mostrar datos")
    print("4. Aplicar comisión")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        cantidad = float(input("Cantidad a agregar: "))
        cuenta.agregar_saldo(cantidad)

    elif opcion == "2":
        cantidad = float(input("Cantidad a retirar: "))
        cuenta.retirar_dinero(cantidad)

    elif opcion == "3":
        cuenta.mostrar_datos()

    elif opcion == "4":
        cuenta.aplicar_comision()
        print("Comisión aplicada correctamente.")

    elif opcion == "5":
        print("Saliendo del sistema...")
        break

    else:
        print("Opción no válida")
```
